# Matplotlib for Machine Learning & Data Science Students

A practical, student-friendly notebook covering the most useful **Matplotlib** commands for data analysis, ML, and model evaluation.

## Learning path
**Figure → Axes → Basic plots → Customization → Multiple plots → Statistical plots → ML visualizations → Model evaluation → Save figures**

Matplotlib is the foundation behind many Python visualization libraries.


```python
# Install if needed:
# !pip install matplotlib pandas numpy scikit-learn

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("Matplotlib version:", plt.matplotlib.__version__)
```

## 1. Basic Line Plot

A line plot is useful for trends over time or ordered observations.

```python
x = [1, 2, 3, 4, 5]
y = [10, 15, 13, 20, 25]

plt.plot(x, y)
plt.show()
```

## 2. Add Title, Labels and Grid

```python
plt.plot(x, y)

plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.show()
```

## 3. Line Styles and Markers

```python
plt.plot(
    x, y,
    marker="o",
    linestyle="--",
    linewidth=2,
    label="Sales"
)

plt.title("Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

## 4. Multiple Lines

```python
months = [1, 2, 3, 4, 5]
sales_a = [10, 15, 13, 20, 25]
sales_b = [8, 12, 16, 18, 23]

plt.plot(months, sales_a, marker="o", label="Product A")
plt.plot(months, sales_b, marker="s", label="Product B")

plt.title("Product Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True)
plt.show()
```

## 5. Bar Chart

Bar charts are useful for comparing categories.

```python
departments = ["IT", "HR", "Finance", "Sales"]
employees = [40, 20, 25, 35]

plt.bar(departments, employees)

plt.title("Employees by Department")
plt.xlabel("Department")
plt.ylabel("Employees")
plt.show()
```

## 6. Horizontal Bar Chart

```python
plt.barh(departments, employees)

plt.title("Employees by Department")
plt.xlabel("Employees")
plt.ylabel("Department")
plt.show()
```

## 7. Grouped Bar Chart

```python
departments = ["IT", "HR", "Finance"]
male = [25, 10, 15]
female = [15, 20, 10]

pos = np.arange(len(departments))
width = 0.35

plt.bar(pos - width/2, male, width, label="Male")
plt.bar(pos + width/2, female, width, label="Female")

plt.xticks(pos, departments)
plt.ylabel("Employees")
plt.title("Employees by Department and Gender")
plt.legend()
plt.show()
```

## 8. Scatter Plot

Scatter plots are useful for studying relationships between two numerical variables.

```python
age = np.array([20, 22, 25, 28, 30, 35, 40, 45])
salary = np.array([35, 38, 45, 50, 55, 65, 75, 90])

plt.scatter(age, salary)

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

## 9. Histogram

Histograms show the distribution of a numerical variable.

```python
rng = np.random.default_rng(42)
salary = rng.normal(60000, 12000, 500)

plt.hist(salary, bins=20)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.show()
```

## 10. Histogram Density

```python
plt.hist(salary, bins=20, density=True)

plt.title("Salary Distribution Density")
plt.xlabel("Salary")
plt.ylabel("Density")
plt.show()
```

## 11. Box Plot

Box plots help identify median, spread, and potential outliers.

```python
data = [45, 50, 52, 55, 58, 60, 62, 65, 100]

plt.boxplot(data)

plt.title("Box Plot")
plt.ylabel("Value")
plt.show()
```

## 12. Pie Chart

Pie charts can show simple proportions. For many categories, bar charts are often easier to read.

```python
labels = ["IT", "HR", "Finance", "Sales"]
values = [40, 20, 25, 35]

plt.pie(
    values,
    labels=labels,
    autopct="%1.1f%%"
)

plt.title("Department Distribution")
plt.show()
```

## 13. Figure and Axes

The object-oriented Matplotlib style is important for professional ML/data-science visualizations.

- `fig` = complete figure
- `ax` = individual plotting area

```python
fig, ax = plt.subplots()

ax.plot(x, y, marker="o")
ax.set_title("Sales")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.grid(True)

plt.show()
```

## 14. Subplots

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot(x, y, marker="o")
axes[0].set_title("Line Plot")

axes[1].bar(departments, employees)
axes[1].set_title("Bar Plot")

plt.tight_layout()
plt.show()
```

## 15. 2 × 2 Subplots

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

axes[0, 0].plot(x, y)
axes[0, 0].set_title("Line")

axes[0, 1].bar(departments, employees)
axes[0, 1].set_title("Bar")

axes[1, 0].scatter(age, salary)
axes[1, 0].set_title("Scatter")

axes[1, 1].hist(salary, bins=15)
axes[1, 1].set_title("Histogram")

plt.tight_layout()
plt.show()
```

## 16. Annotating a Point

```python
x = np.array([1,2,3,4,5])
y = np.array([10,15,13,20,25])

plt.plot(x, y, marker="o")

plt.annotate(
    "Highest",
    xy=(5, 25),
    xytext=(3.5, 22),
    arrowprops={"arrowstyle": "->"}
)

plt.show()
```

## 17. Axis Limits

```python
plt.plot(x, y)

plt.xlim(1, 5)
plt.ylim(0, 30)

plt.title("Controlled Axis Range")
plt.show()
```

## 18. Twin Y-Axis

Useful when two variables have very different scales.

```python
months = np.arange(1, 7)
sales = [20, 25, 30, 35, 40, 45]
profit = [2, 3, 4, 5, 6, 8]

fig, ax1 = plt.subplots()

ax1.plot(months, sales, marker="o")
ax1.set_xlabel("Month")
ax1.set_ylabel("Sales")

ax2 = ax1.twinx()
ax2.plot(months, profit, marker="s", linestyle="--")
ax2.set_ylabel("Profit")

plt.title("Sales and Profit")
plt.show()
```

## 19. Correlation Heatmap with Matplotlib

A heatmap can be created directly with `imshow()`.

```python
data = np.array([
    [1.0, 0.8, 0.2],
    [0.8, 1.0, 0.4],
    [0.2, 0.4, 1.0]
])

plt.imshow(data, aspect="auto")
plt.colorbar()

plt.xticks(range(3), ["Age", "Salary", "Experience"])
plt.yticks(range(3), ["Age", "Salary", "Experience"])

plt.title("Correlation Matrix")
plt.show()
```

## 20. ML: Actual vs Predicted

```python
actual = np.array([100, 120, 130, 150, 170])
predicted = np.array([105, 115, 128, 145, 175])

plt.scatter(actual, predicted)

# Ideal prediction line
minimum = min(actual.min(), predicted.min())
maximum = max(actual.max(), predicted.max())

plt.plot([minimum, maximum], [minimum, maximum], linestyle="--")

plt.xlabel("Actual")
plt.ylabel("Predicted")
plt.title("Actual vs Predicted")
plt.grid(True)
plt.show()
```

## 21. ML: Residual Plot

Residual = Actual − Predicted

```python
residuals = actual - predicted

plt.scatter(predicted, residuals)
plt.axhline(0, linestyle="--")

plt.xlabel("Predicted")
plt.ylabel("Residual")
plt.title("Residual Plot")
plt.grid(True)
plt.show()
```

## 22. ML: Confusion Matrix Visualization

```python
from sklearn.metrics import confusion_matrix

y_true = [0, 0, 0, 1, 1, 1, 1, 0]
y_pred = [0, 1, 0, 1, 1, 0, 1, 0]

cm = confusion_matrix(y_true, y_pred)

plt.imshow(cm)
plt.colorbar()

plt.xticks([0, 1], ["Pred 0", "Pred 1"])
plt.yticks([0, 1], ["Actual 0", "Actual 1"])

for i in range(cm.shape[0]):
    for j in range(cm.shape[1]):
        plt.text(j, i, cm[i, j], ha="center", va="center")

plt.title("Confusion Matrix")
plt.show()
```

## 23. ML: Training Loss Curve

```python
epochs = np.arange(1, 11)
train_loss = [0.90, 0.70, 0.55, 0.45, 0.37, 0.31, 0.27, 0.24, 0.22, 0.20]
val_loss = [0.95, 0.76, 0.62, 0.54, 0.49, 0.47, 0.48, 0.50, 0.53, 0.56]

plt.plot(epochs, train_loss, marker="o", label="Training Loss")
plt.plot(epochs, val_loss, marker="s", label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.grid(True)
plt.show()
```

## 24. Feature Importance

```python
features = ["Age", "Income", "Experience", "Education"]
importance = [0.15, 0.45, 0.30, 0.10]

plt.barh(features, importance)

plt.xlabel("Importance")
plt.title("Feature Importance")
plt.show()
```

## 25. Save a Figure

```python
plt.plot(x, y, marker="o")
plt.title("Sales Trend")

# Save as PNG
plt.savefig("sales_trend.png", dpi=300, bbox_inches="tight")

plt.show()
```

## 26. Style Basics

Matplotlib can be configured with labels, legends, grids, markers, line styles, figure size, and transparency.

For teaching, focus first on understanding the data and the correct chart type rather than decorative styling.

```python
fig, ax = plt.subplots(figsize=(8, 5))

ax.plot(
    x, y,
    marker="o",
    linewidth=2,
    alpha=0.8,
    label="Sales"
)

ax.set_title("Professional-Style Sales Plot")
ax.set_xlabel("Month")
ax.set_ylabel("Sales")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

# Matplotlib Quick Reference

| Purpose | Command |
|---|---|
| Line | `plt.plot()` |
| Bar | `plt.bar()` |
| Horizontal bar | `plt.barh()` |
| Scatter | `plt.scatter()` |
| Histogram | `plt.hist()` |
| Box plot | `plt.boxplot()` |
| Pie | `plt.pie()` |
| Figure | `plt.figure()` |
| Subplots | `plt.subplots()` |
| Title | `plt.title()` / `ax.set_title()` |
| X label | `plt.xlabel()` / `ax.set_xlabel()` |
| Y label | `plt.ylabel()` / `ax.set_ylabel()` |
| Legend | `plt.legend()` / `ax.legend()` |
| Grid | `plt.grid()` / `ax.grid()` |
| Limits | `plt.xlim()`, `plt.ylim()` |
| Annotation | `plt.annotate()` |
| Save | `plt.savefig()` |
| Show | `plt.show()` |
