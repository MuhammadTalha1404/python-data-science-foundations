# Seaborn for Machine Learning & Data Science Students

A practical notebook covering the most useful **Seaborn** visualizations for EDA, statistics, ML, and data storytelling.

Seaborn is built on Matplotlib and provides a higher-level interface for attractive statistical graphics.


```python
# Install if needed:
# !pip install seaborn pandas numpy matplotlib scikit-learn

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Seaborn version:", sns.__version__)

# Load an attractive default style
sns.set_theme()
```

## 1. Create a Sample Dataset

```python
rng = np.random.default_rng(42)

df = pd.DataFrame({
    "Age": rng.integers(20, 55, 200),
    "Salary": rng.normal(65000, 15000, 200).round(0),
    "Experience": rng.integers(1, 20, 200),
    "Department": rng.choice(["IT", "HR", "Finance", "Sales"], 200),
    "Gender": rng.choice(["Male", "Female"], 200),
    "Performance": rng.normal(70, 12, 200).round(1)
})

display(df.head())
```

## 2. Scatter Plot

Use scatter plots to investigate relationships between two numerical variables.

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Salary"
)

plt.title("Age vs Salary")
plt.show()
```

## 3. Scatter Plot with Hue

`hue` adds a categorical dimension using different visual groups.

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Salary",
    hue="Department"
)

plt.title("Salary by Age and Department")
plt.show()
```

## 4. Scatter Plot with Size and Style

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Salary",
    hue="Gender",
    size="Experience",
    style="Gender"
)

plt.title("Multi-Dimensional Scatter Plot")
plt.show()
```

## 5. Line Plot

Useful for trends over an ordered variable such as time.

```python
monthly = pd.DataFrame({
    "Month": range(1, 13),
    "Sales": [100, 120, 115, 140, 155, 160, 175, 170, 190, 205, 220, 240]
})

sns.lineplot(
    data=monthly,
    x="Month",
    y="Sales",
    marker="o"
)

plt.title("Monthly Sales")
plt.show()
```

## 6. Bar Plot

Seaborn's `barplot()` commonly shows an aggregate such as the mean for each category, with uncertainty information.

```python
sns.barplot(
    data=df,
    x="Department",
    y="Salary"
)

plt.title("Average Salary by Department")
plt.show()
```

## 7. Count Plot

Count plots show how many observations belong to each category.

```python
sns.countplot(
    data=df,
    x="Department"
)

plt.title("Employees by Department")
plt.show()
```

## 8. Count Plot with Hue

```python
sns.countplot(
    data=df,
    x="Department",
    hue="Gender"
)

plt.title("Employees by Department and Gender")
plt.show()
```

## 9. Histogram

`histplot()` shows the distribution of a numerical variable.

```python
sns.histplot(
    data=df,
    x="Salary",
    bins=20
)

plt.title("Salary Distribution")
plt.show()
```

## 10. Histogram with KDE

```python
sns.histplot(
    data=df,
    x="Salary",
    bins=20,
    kde=True
)

plt.title("Salary Distribution with KDE")
plt.show()
```

## 11. KDE Plot

KDE estimates the shape of a continuous probability distribution.

```python
sns.kdeplot(
    data=df,
    x="Salary",
    fill=True
)

plt.title("Salary Density")
plt.show()
```

## 12. KDE by Category

```python
sns.kdeplot(
    data=df,
    x="Salary",
    hue="Gender",
    fill=True,
    common_norm=False
)

plt.title("Salary Distribution by Gender")
plt.show()
```

## 13. Box Plot

Excellent for comparing distributions and spotting potential outliers.

```python
sns.boxplot(
    data=df,
    x="Department",
    y="Salary"
)

plt.title("Salary Distribution by Department")
plt.show()
```

## 14. Box Plot with Hue

```python
sns.boxplot(
    data=df,
    x="Department",
    y="Salary",
    hue="Gender"
)

plt.title("Salary by Department and Gender")
plt.show()
```

## 15. Violin Plot

A violin plot combines aspects of a box plot and a density plot.

```python
sns.violinplot(
    data=df,
    x="Department",
    y="Salary"
)

plt.title("Salary Distribution by Department")
plt.show()
```

## 16. Strip Plot

Shows individual observations for each category.

```python
sns.stripplot(
    data=df,
    x="Department",
    y="Salary",
    jitter=True
)

plt.title("Individual Salary Observations")
plt.show()
```

## 17. Swarm Plot

A swarm plot attempts to show individual observations without overlapping them.

```python
small_df = df.head(80)

sns.swarmplot(
    data=small_df,
    x="Department",
    y="Salary"
)

plt.title("Salary Observations")
plt.show()
```

## 18. Regression Plot

`regplot()` adds a regression line to a scatter plot.

```python
sns.regplot(
    data=df,
    x="Experience",
    y="Salary"
)

plt.title("Experience vs Salary")
plt.show()
```

## 19. Regression Plot with Confidence Interval

```python
sns.regplot(
    data=df,
    x="Age",
    y="Performance",
    ci=95
)

plt.title("Age vs Performance")
plt.show()
```

## 20. Heatmap

Heatmaps are commonly used for correlation matrices.

```python
corr = df[["Age", "Salary", "Experience", "Performance"]].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

## 21. Heatmap with Colorbar Settings

```python
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    linewidths=0.5,
    cbar=True
)

plt.title("Detailed Correlation Heatmap")
plt.show()
```

## 22. Pair Plot

`pairplot()` creates multiple pairwise relationships and distributions.

This is extremely useful for quick EDA on small-to-medium datasets.

```python
pair_df = df[["Age", "Salary", "Experience", "Performance", "Gender"]]

sns.pairplot(
    pair_df,
    hue="Gender"
)

plt.show()
```

## 23. Joint Plot

Shows the relationship between two variables together with their marginal distributions.

```python
sns.jointplot(
    data=df,
    x="Age",
    y="Salary",
    kind="scatter"
)

plt.show()
```

## 24. Joint KDE Plot

```python
sns.jointplot(
    data=df,
    x="Age",
    y="Salary",
    kind="kde",
    fill=True
)

plt.show()
```

## 25. Joint Regression Plot

```python
sns.jointplot(
    data=df,
    x="Experience",
    y="Salary",
    kind="reg"
)

plt.show()
```

## 26. Catplot

`catplot()` provides a figure-level interface for categorical plots.

```python
sns.catplot(
    data=df,
    x="Department",
    y="Salary",
    kind="box",
    col="Gender"
)

plt.show()
```

## 27. FacetGrid

FacetGrid creates multiple plots for subsets of the data.

```python
g = sns.FacetGrid(
    df,
    col="Gender",
    row="Department"
)

g.map_dataframe(
    sns.scatterplot,
    x="Age",
    y="Salary"
)

plt.show()
```

## 28. Hue for a Third Variable

```python
sns.scatterplot(
    data=df,
    x="Experience",
    y="Performance",
    hue="Department"
)

plt.title("Experience vs Performance")
plt.show()
```

## 29. Categorical vs Numerical Analysis

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.boxplot(
    data=df,
    x="Department",
    y="Performance",
    ax=axes[0]
)

sns.violinplot(
    data=df,
    x="Department",
    y="Performance",
    ax=axes[1]
)

axes[0].set_title("Box Plot")
axes[1].set_title("Violin Plot")

plt.tight_layout()
plt.show()
```

## 30. Distribution Comparison

```python
sns.histplot(
    data=df,
    x="Performance",
    hue="Gender",
    kde=True,
    element="step",
    stat="density",
    common_norm=False
)

plt.title("Performance Distribution")
plt.show()
```

## 31. Outlier Analysis

```python
sns.boxplot(
    data=df,
    x="Salary"
)

plt.title("Salary Outlier Analysis")
plt.show()
```

## 32. Correlation Analysis for ML

Correlation can help understand linear relationships between numerical features.

Remember: correlation does **not** prove causation.

```python
numeric_cols = ["Age", "Salary", "Experience", "Performance"]

corr = df[numeric_cols].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    center=0
)

plt.title("ML Feature Correlation")
plt.show()
```

## 33. ML: Actual vs Predicted

Seaborn can create attractive model evaluation plots.

```python
actual = np.array([100, 120, 130, 150, 170])
predicted = np.array([105, 115, 128, 145, 175])

evaluation_df = pd.DataFrame({
    "Actual": actual,
    "Predicted": predicted
})

sns.scatterplot(
    data=evaluation_df,
    x="Actual",
    y="Predicted",
    s=100
)

minimum = min(actual.min(), predicted.min())
maximum = max(actual.max(), predicted.max())

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--"
)

plt.title("Actual vs Predicted")
plt.show()
```

## 34. ML: Residual Analysis

```python
evaluation_df["Residual"] = (
    evaluation_df["Actual"] - evaluation_df["Predicted"]
)

sns.scatterplot(
    data=evaluation_df,
    x="Predicted",
    y="Residual",
    s=100
)

plt.axhline(0, linestyle="--")
plt.title("Residual Plot")
plt.show()
```

## 35. ML: Confusion Matrix Heatmap

```python
from sklearn.metrics import confusion_matrix

y_true = [0, 0, 0, 1, 1, 1, 1, 0]
y_pred = [0, 1, 0, 1, 1, 0, 1, 0]

cm = confusion_matrix(y_true, y_pred)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["Pred 0", "Pred 1"],
    yticklabels=["Actual 0", "Actual 1"]
)

plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

## 36. ML: Model Feature Importance

```python
features = ["Age", "Income", "Experience", "Education"]
importance = [0.15, 0.45, 0.30, 0.10]

importance_df = pd.DataFrame({
    "Feature": features,
    "Importance": importance
}).sort_values("Importance", ascending=False)

sns.barplot(
    data=importance_df,
    x="Importance",
    y="Feature"
)

plt.title("Feature Importance")
plt.show()
```

## 37. ML: Training and Validation Loss

```python
epochs = np.arange(1, 11)

loss_df = pd.DataFrame({
    "Epoch": epochs,
    "Training Loss": [0.90,0.70,0.55,0.45,0.37,0.31,0.27,0.24,0.22,0.20],
    "Validation Loss": [0.95,0.76,0.62,0.54,0.49,0.47,0.48,0.50,0.53,0.56]
})

sns.lineplot(
    data=loss_df,
    x="Epoch",
    y="Training Loss",
    marker="o",
    label="Training"
)

sns.lineplot(
    data=loss_df,
    x="Epoch",
    y="Validation Loss",
    marker="s",
    label="Validation"
)

plt.title("Training vs Validation Loss")
plt.show()
```

## 38. Change Plot Theme

Seaborn provides convenient themes for visualization.

```python
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=df,
    x="Age",
    y="Salary",
    hue="Department"
)

plt.title("Seaborn Theme Example")
plt.show()
```

## 39. Common Seaborn Styles

Try:
- `white`
- `dark`
- `whitegrid`
- `darkgrid`
- `ticks`

```python
for style in ["whitegrid", "darkgrid", "ticks"]:
    sns.set_theme(style=style)
    sns.boxplot(data=df, x="Department", y="Salary")
    plt.title(f"Style: {style}")
    plt.show()
```

## 40. Save a Seaborn Figure

```python
sns.set_theme(style="whitegrid")

sns.scatterplot(
    data=df,
    x="Age",
    y="Salary",
    hue="Department"
)

plt.title("Salary by Age")
plt.savefig(
    "seaborn_salary_plot.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

## EDA mental model

`Dataset → Distribution → Categories → Relationships → Correlation → Outliers → Patterns → ML Insights`

### Important distinction

**Matplotlib:** lower-level, highly flexible plotting foundation.

**Seaborn:** higher-level statistical visualization built on Matplotlib.

In real projects, it is common to use **both together**.
